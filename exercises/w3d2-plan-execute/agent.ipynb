# Week 3 Day 2：Plan-and-Execute / LLMCompiler

目标：把复杂任务先编译成带依赖关系的计划，再按 DAG 分批执行。重点不是“列 todo”，而是让计划具备输入、输出、依赖和可重跑边界。

## 新手导读：Plan-and-Execute 是“先画施工图，再动手”

ReAct 是边观察边行动，适合短任务和信息不确定的场景。Plan-and-Execute 适合多步骤、需要审计、可以提前拆解的任务。

本节关键词：

- Plan：把用户目标拆成步骤，每步有输入、输出、依赖。
- Execute：按依赖顺序执行步骤，而不是靠模型每轮临时决定。
- DAG：有向无环图，表示哪些步骤必须先做，哪些可以并行。
- LLMCompiler 思路：把自然语言任务“编译”成可调度的执行图。

阅读顺序建议：

1. 先看 `build_plan` 输出的结构，不要急着看执行。
2. 再看 `execute_plan`，理解依赖满足后才能执行。
3. 看 `plan_to_mermaid`，把计划图可视化，帮助发现依赖设计是否合理。
4. 最后跑 `self_test`，确认计划和执行结果可验证。

常见误区：

- 计划不是自然语言 todo list；工程计划必须有依赖和产物。
- Plan-and-Execute 不一定比 ReAct 好。任务变化快、需要实时观察时，ReAct 更灵活。
- 可并行不等于一定并行；先显式建 DAG，才有优化空间。


## ReAct vs Plan-and-Execute

- ReAct：每一步都在观察后决定下一步，适合短任务、环境反馈频繁的任务。
- Plan-and-Execute：先做全局规划，再执行，适合多步骤、可审计、可并行的任务。
- LLMCompiler：进一步把计划变成依赖图，识别可以并行执行的节点。
