# Week 3 Day 2：Plan-and-Execute / LLMCompiler

目标：把复杂任务先编译成带依赖关系的计划，再按 DAG 分批执行。重点不是“列 todo”，而是让计划具备输入、输出、依赖和可重跑边界。

## ReAct vs Plan-and-Execute

- ReAct：每一步都在观察后决定下一步，适合短任务、环境反馈频繁的任务。
- Plan-and-Execute：先做全局规划，再执行，适合多步骤、可审计、可并行的任务。
- LLMCompiler：进一步把计划变成依赖图，识别可以并行执行的节点。
