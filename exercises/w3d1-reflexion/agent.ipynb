# Week 3 Day 1：Reflexion + Self-Refine

目标：把“回答失败”变成可复用的文字记忆，并用 `generator -> critic -> refiner` 的闭环改进下一轮输出。

## 新手导读：Reflexion 是“失败后写下可执行经验”

这一节是离线模拟，不依赖真实 LLM。重点不是 API，而是理解一种 Agent 改进模式：

- 第一次生成：Agent 按当前策略尝试解决任务。
- 批判/评估：找出答案哪里不满足要求。
- 反思：把失败原因写成下次可用的文字记忆。
- 再生成：下一轮显式读取反思，避免重复犯错。

Reflexion 和 Self-Refine 的差别：

- Self-Refine 更像“同一轮内自我修改”：生成 → 批判 → 改写。
- Reflexion 更强调“跨轮学习”：把失败轨迹沉淀成记忆，下次再用。

阅读顺序建议：

1. 先运行 `run_reflexion`，看 trace 里每一轮如何变化。
2. 再看 `print_trace`，把轨迹当成调试日志读。
3. 最后运行 `self_test`，确认离线规则能稳定复现。

常见误区：

- 反思不是写“我要更认真”，而是写“下次具体检查 X、补 Y、避免 Z”。
- 反思必须可验证，否则只是把更多废话塞进上下文。
- Reflexion 不能替代真实评测；它依赖 critic 是否能发现问题。


## 核心区别

- ReAct 解决的是单轮任务中的 `Reasoning + Acting`。
- Reflexion 解决的是跨轮学习：失败后把原因写入 memory。
- Self-Refine 解决的是单模型自我批判：生成、批判、修订分成三个角色。
